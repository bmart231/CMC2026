# Tensors

Tensors are a data structure—a way of storing numbers. They are very similar to arrays, which you may have encountered before (such as the ndarrays in NumPy). A tensor can be as simple as a one-dimensional list of numbers:

$$
\begin{bmatrix} 1 & 2 & 3 \end{bmatrix}
$$

But a tensor can also be a 2-dimensional array of numbers:

$$
\begin{bmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 7 & 8 & 9 \end{bmatrix}
$$

Or it can have 3, 4, or any other number of dimensions. You might talk about a 1x3 tensor (like the first example above) or a 3x3 tensor (like the second example) or an Nx1x32x32 tensor (which we'll see in a later notebook). So, at least as far as the field of machine learning is concerned, a tensor is just a big block of numbers—a multidimensional array.

PyTorch uses tensors to store the data that make up the input to a neural network model (e.g. images for classification), and also uses them to store the parameters that define the model (its weights and biases).

--------------

```python
import torch
import numpy as np
```

## 1. Initializing Tensors

There are several ways to initialize a tensor—let's look at some examples:

**1.1 Directly from data**

Tensors can be created directly from data. The data type is automatically inferred.

```python
data = [[1, 2], [3, 4]]
data_tensor = torch.tensor(data)
print(data_tensor, data_tensor.dtype)
```

**1.2 From a NumPy array**

Tensors can be created from NumPy arrays. Note that the new tensor and the NumPy array will share the same memory.

```python
np_array = np.array(data)
data_tensor = torch.from_numpy(np_array)
print(data_tensor)
```

**1.3 With random or constant values**

``shape`` is a tuple of tensor dimensions.



```python
shape = (2, 3)
rand_tensor = torch.rand(shape)
ones_tensor = torch.ones(shape)
zeros_tensor = torch.zeros(shape)

print(f"Random Tensor: \n {rand_tensor} \n")
print(f"Ones Tensor: \n {ones_tensor} \n")
print(f"Zeros Tensor: \n {zeros_tensor}")
```

**1.4 Using another tensor as a template**

The new tensor retains the properties (shape, datatype) of the argument tensor, unless explicitly overridden.



```python
data_tensor = torch.tensor([[1, 2], [3, 4]])

print(f"Data Tensor: \n {data_tensor} \n")

x_ones = torch.ones_like(data_tensor) # retains the properties of data_tensor
print(f"Ones Tensor: \n {x_ones} \n")

x_rand = torch.rand_like(data_tensor, dtype=torch.float) # overrides the datatype of data_tensor
print(f"Random Tensor: \n {x_rand} \n")
```

--------------




## 2. Tensor Attributes

Tensor attributes describe their shape, datatype, and the device on which they are stored.



```python
tensor = torch.rand(3, 4)

print(f"Tensor: \n {tensor} \n")

#Here are 3 common attributes:
print(f"Shape of tensor: {tensor.shape}")
print(f"Datatype of tensor: {tensor.dtype}")
print(f"Device tensor is stored on: {tensor.device}")
#a device is the piece of hardware whose memory holds the tensor and whose processor runs operations on it.
#Examples:
#cpu — normal system RAM, computed by the CPU
#cuda:0 — the first NVIDIA GPU (cuda:1 would be the second)
#mps — Apple's Metal GPU backend, which is what you'd get on your Mac
```

## 3. Tensor Operations

A full list of tensor operations can be found [here](https://pytorch.org/docs/stable/torch.html).

**3.1 Arithmetic (element-wise)**

You can use the usual +, -, \*, / symbols

```python
x = torch.randint(10, (3, 2))
y = torch.randint(10, (3, 2))
print(x, '\n')
print(y, '\n')
z = x + y
print(z, '\n')
w = x * y
print(w, '\n')
```

**3.2 Matrix multiplication**

```python
x = torch.tensor([[0, 1], [-1, 0]])
y = torch.tensor([[4, 0], [0, 4]])
print(x, '\n')
print(y, '\n')
z = x @ y
print(z)
```

**3.3 Standard numpy-like indexing and slicing**

```python
tensor = torch.zeros(3, 3)
print(tensor, '\n')
tensor[0, 0] = 3
tensor[:, 2] = 5
print(tensor)
```

**3.4 In-place operations**

Operations that have a ``_`` suffix are in-place. For example: ``x.copy_(y)``, ``x.t_()``, will change ``x``.

```python
tensor = torch.zeros(3, 3)
tensor[0, 0] = 3
tensor[:, 2] = 5

print(tensor, "\n")

tensor.add_(20)
print(tensor, "\n")

tensor.sqrt_()
print(tensor, "\n")
```

<div class="alert alert-info"><h4>Note</h4><p>In-place operations save some memory, but can potentially overwrite values required to compute gradients, so their use is discouraged.</p></div>

**3.5 Joining tensors**

You can use ``torch.cat`` to concatenate a sequence of tensors along a given dimension. See also [torch.stack](https://pytorch.org/docs/stable/generated/torch.stack.html), another operation that is subtly different from ``torch.cat``.

```python
t1 = torch.cat([tensor, tensor], dim=1)
print(t1)
```

**3.6 `torch.stack` vs `torch.cat`**

Both combine several tensors into one, but they do it differently. `torch.cat` joins tensors along an *existing* dimension, so the output has the same number of dimensions as the inputs. `torch.stack` adds a *new* dimension, so the inputs must all have the exact same shape, and the output has one more dimension than they do.

```python
a = torch.zeros(2, 3)
b = torch.ones(2, 3)

print(f"a: {a}")
print(f"b: {b}", '\n')
```

```python
# cat: joins along an existing dimension, so the inputs' number of dims is preserved
print(f"cat along dim=0: {torch.cat([a, b], dim=0).shape}")
print(torch.cat([a, b], dim=0), '\n')

print(f"cat along dim=1: {torch.cat([a, b], dim=1).shape}")
print(torch.cat([a, b], dim=1), '\n')
```

```python
# stack: adds a new dimension, so the inputs must have the exact same shape
print(f"stack along dim=0: {torch.stack([a, b], dim=0).shape}")
print(torch.stack([a, b], dim=0), '\n')

print(f"stack along dim=1: {torch.stack([a, b], dim=1).shape}")
print(torch.stack([a, b], dim=1), '\n')
```

**3.7 Changing device**

Typically, tensor operations will be faster on a GPU (if supported) than a CPU.

```python
import time
```

```python
x = torch.rand(8_000, 8_000)
y = torch.rand(8_000, 8_000)
start_time = time.time()

# this should take several seconds to run:
z = x @ y

end_time = time.time()
elapsed_time = end_time - start_time
print(np.round(elapsed_time, 3), " seconds")
```

```python
# move our tensor to the GPU if available

# CUDA: Nvidia GPUs
if torch.cuda.is_available():
    x = x.to('cuda')
    y = y.to('cuda')
    print("CUDA is available. Tensors moved to GPU.")

# MPS: Mac GPUs
if torch.backends.mps.is_available():
    x = x.to('mps')
    y = y.to('mps')
    print("MPS is available. Tensors moved to GPU.")

print(f"Tensor is stored on: {x.device}")
```

```python
# if the tensors were moved to a GPU, this should be faster
z = 0
start_time = time.time()
z = x @ y

end_time = time.time()
elapsed_time = end_time - start_time
print(np.round(elapsed_time, 3), " seconds")
```
